Step 4.1 — Read the cleaned dataset

```python
df.head()
```

Step 4.2 — Check the event distribution again

Step 4.3 — Create session-level funnel flags

```python
session_funnel = df.groupby('user_session').agg(
    user_id=('user_id', 'first'),
    viewed=('event_type', lambda x: (x == 'view').any()),
    added_to_cart=('event_type', lambda x: (x == 'cart').any()),
    purchased=('event_type', lambda x: (x == 'purchase').any())
).reset_index()
```

```python
session_funnel.head()
```

Step 4.4 — Count sessions at each funnel stage

```python
total_sessions = session_funnel['user_session'].nunique()

view_sessions = session_funnel['viewed'].sum()

cart_sessions = session_funnel['added_to_cart'].sum()

purchase_sessions = session_funnel['purchased'].sum()

print("Total Sessions:", total_sessions)
print("View Sessions:", view_sessions)
print("Cart Sessions:", cart_sessions)
print("Purchase Sessions:", purchase_sessions)
```

Step 4.5 — Calculate conversion rates

```python
view_to_cart = (cart_sessions / view_sessions) * 100
cart_to_purchase = (purchase_sessions / cart_sessions) * 100
overall_conversion = (purchase_sessions / view_sessions) * 100

print("View → Cart Conversion:", round(view_to_cart, 2), "%")
print("Cart → Purchase Conversion:", round(cart_to_purchase, 2), "%")
print("Overall Conversion:", round(overall_conversion, 2), "%")
```

```python
### and drop off:
overall_dropoff = 100 - overall_conversion

print("Overall Drop-off:", round(overall_dropoff, 2), "%")
```

STEP 5 — Find the Biggest Drop-Off
We now want to answer our main business question:

At which stage are we losing the most potential customers?

Step 5.1 — Calculate the number of sessions lost at each stage

```python
view_to_cart_dropoff = view_sessions - cart_sessions
cart_to_purchase_dropoff = cart_sessions - purchase_sessions

print("View → Cart Drop-off:", view_to_cart_dropoff)
print("Cart → Purchase Drop-off:", cart_to_purchase_dropoff)
```

Step 5.2 — Calculate drop-off percentages

```python
view_to_cart_dropoff_pct = ((view_sessions - cart_sessions) / view_sessions) * 100

cart_to_purchase_dropoff_pct = ((cart_sessions - purchase_sessions) / cart_sessions) * 100

print("View → Cart Drop-off:", round(view_to_cart_dropoff_pct, 2), "%")
print("Cart → Purchase Drop-off:", round(cart_to_purchase_dropoff_pct, 2), "%")
```

Step 5.3 — Automatically identify the biggest drop-off

```python
dropoffs = {
    'View → Cart': view_to_cart_dropoff_pct,
    'Cart → Purchase': cart_to_purchase_dropoff_pct
}

largest_dropoff_stage = max(dropoffs, key=dropoffs.get)

print("Largest Drop-off Stage:", largest_dropoff_stage)
print("Drop-off Rate:", round(dropoffs[largest_dropoff_stage], 2), "%")
```

Step 5.4 — Create a small funnel table

```python
funnel = pd.DataFrame({
    'Stage': ['View', 'Cart', 'Purchase'],
    'Sessions': [view_sessions, cart_sessions, purchase_sessions]
})

funnel
```

Step 6.6 — Calculate drop-off

STEP 10 — User & Session Behavior Analysis
Our funnel tells us where the drop-off happens. Now we want to understand how users behave.

The first thing we'll measure is:

How many sessions does each user have?

This helps us distinguish one-time visitors from users who return multiple times.

Step 10.1 — Sessions per User

```python
sessions_per_user = (
    df.groupby('user_id')['user_session']
    .nunique()
    .reset_index(name='session_count')
)

sessions_per_user.head()
```

Step 10.2 — Understand the distribution

```python
sessions_per_user['session_count'].describe()
```

This tells us:

-average sessions per user
-minimum
-maximum
-median
-quartiles

Step 10.3 — Segment users by number of sessions

```python
def classify_user(x):
    if x == 1:
        return '1 Session'
    elif x <= 3:
        return '2-3 Sessions'
    else:
        return '4+ Sessions'

sessions_per_user['user_segment'] = (
    sessions_per_user['session_count']
    .apply(classify_user)
)
```

```python
sessions_per_user['user_segment'].value_counts()
```

Step 10.4 — Connect user behavior to purchasing

Do users who return more often have a higher likelihood of purchasing?

```python
user_purchase = (
    df.groupby('user_id')['event_type']
    .apply(lambda x: (x == 'purchase').any())
    .reset_index(name='has_purchased')
)
```

```python
user_behavior = sessions_per_user.merge(
    user_purchase,
    on='user_id',
    how='left'
)
```

```python
user_behavior.head()
```

Step 10.5 — Calculate purchase rate by user segment

```python
user_segment_analysis = (
    user_behavior
    .groupby('user_segment')
    .agg(
        Users=('user_id', 'nunique'),
        Purchasing_Users=('has_purchased', 'sum')
    )
    .reset_index()
)

user_segment_analysis['Purchase_Rate_%'] = (
    user_segment_analysis['Purchasing_Users']
    / user_segment_analysis['Users']
    * 100
)

user_segment_analysis
```

There is a clear relationship between repeat sessions and purchasing:

1-session users → 6.17%
2–3-session users → 14.41%
4+ sessions → 31.20%

So users with more sessions show substantially higher purchase rates.

```python
df['hour'] = df['event_datetime'].dt.hour
```

```python
df[['event_time', 'event_datetime', 'hour']].head()
```

Step 16.2 — Create time-of-day bands

```python
def time_band(hour):
    if 0 <= hour < 6:
        return 'Night (00-06)'
    elif 6 <= hour < 12:
        return 'Morning (06-12)'
    elif 12 <= hour < 18:
        return 'Afternoon (12-18)'
    else:
        return 'Evening (18-24)'

df['time_band'] = df['hour'].apply(time_band)
```

```python
df['time_band'].value_counts()
```

Step 16.3 — Create sequential time funnel

```python
time_events = (
    df.groupby(['user_session', 'time_band', 'event_type'])['event_datetime']
    .min()
    .reset_index()
)
```

```python
time_first = (
    time_events
    .pivot_table(
        index=['user_session', 'time_band'],
        columns='event_type',
        values='event_datetime',
        aggfunc='min'
    )
    .reset_index()
)
```

Step 16.4 — Create funnel flags

```python
time_first['viewed'] = time_first['view'].notna()

time_first['cart_after_view'] = (
    time_first['view'].notna() &
    time_first['cart'].notna() &
    (time_first['cart'] > time_first['view'])
)

time_first['purchase_after_cart'] = (
    time_first['cart_after_view'] &
    time_first['purchase'].notna() &
    (time_first['purchase'] > time_first['cart'])
)
```

Step 16.5 — Calculate time-band performance

```python
time_analysis = (
    time_first
    .groupby('time_band')
    .agg(
        View_Sessions=('viewed', 'sum'),
        Cart_Sessions=('cart_after_view', 'sum'),
        Purchase_Sessions=('purchase_after_cart', 'sum')
    )
    .reset_index()
)
```

```python
time_analysis['View_to_Cart_%'] = (
    time_analysis['Cart_Sessions']
    / time_analysis['View_Sessions']
    * 100
)

time_analysis['Cart_to_Purchase_%'] = (
    time_analysis['Purchase_Sessions']
    / time_analysis['Cart_Sessions'].replace(0, pd.NA)
    * 100
)

time_analysis['View_to_Purchase_%'] = (
    time_analysis['Purchase_Sessions']
    / time_analysis['View_Sessions']
    * 100
)
```
